#### Step 1: store the raw data


```python
# Raw product data stored as text (strings) exactly as exported by the system
notebook_units_text = "120"
notebook_selling_price_text = "45"
notebook_cost_price_text = "30"

pen_pack_units_text = "200"
pen_pack_selling_price_text = "20"
pen_pack_cost_price_text = "12"

school_bag_units_text = "35"
school_bag_selling_price_text = "650"
school_bag_cost_price_text = "480"

# Shop-level values stored directly as numbers (floats).
# Why: Unlike the system export, these are manual business targets/overhead.
# Storing them directly as floats avoids unnecessary type conversion later and allows decimals (paise).
monthly_expenses = 6000.0
monthly_profit_target = 5000.0
```

#### Step 2: convert to numbers


```python
# Convert units to integers (int) because you cannot sell a fraction of a physical item.
# Convert prices to floats (float) so the code won't break if prices later include paise (e.g., 45.50).
# Using clean names (without '_text') keeps raw data separate from usable numbers.

# Notebook
notebook_units = int(notebook_units_text)
notebook_selling_price = float(notebook_selling_price_text)
notebook_cost_price = float(notebook_cost_price_text)

# Pen pack
pen_pack_units = int(pen_pack_units_text)
pen_pack_selling_price = float(pen_pack_selling_price_text)
pen_pack_cost_price = float(pen_pack_cost_price_text)

# School bag
school_bag_units = int(school_bag_units_text)
school_bag_selling_price = float(school_bag_selling_price_text)
school_bag_cost_price = float(school_bag_cost_price_text)

# ==========================================
# Print the type of every converted variable
# ==========================================

# Notebook types
print("\nnotebook_units:", type(notebook_units),"\nnotebook_selling_price:", type(notebook_selling_price), "\nnotebook_cost_price:", type(notebook_cost_price))
# Pen pack types
print("\npen_pack_units:", type(pen_pack_units),"\npen_pack_selling_price:", type(pen_pack_selling_price), "\npen_pack_cost_price:", type(pen_pack_cost_price))
# School bag types
print("\nschool_bag_units:", type(school_bag_units),"\nschool_bag_selling_price:", type(school_bag_selling_price),"\nschool_bag_cost_price:", type(school_bag_cost_price))
```

#### Step 3: calculate per product


```python

# --- Notebook ---
# Multiply units by selling price to find total money earned before deducting any costs.
notebook_revenue = notebook_units * notebook_selling_price
# Multiply units by cost price to find total direct product cost for all notebooks sold.
notebook_total_cost = notebook_units * notebook_cost_price
# Subtract total cost from revenue to see how much profit notebooks alone generated.
notebook_profit = notebook_revenue - notebook_total_cost
# Divide profit by revenue (not cost) to measure what percentage of each rupee in sales is profit.
notebook_profit_margin = (notebook_profit / notebook_revenue) * 100
# Subtract cost price from selling price to see the profit earned on a single notebook.
notebook_profit_per_unit = notebook_selling_price - notebook_cost_price

# --- Pen pack ---
# Multiply units by selling price to find gross sales income from pen packs.
pen_pack_revenue = pen_pack_units * pen_pack_selling_price
# Multiply units by cost price to find total purchase cost of the pen packs sold.
pen_pack_total_cost = pen_pack_units * pen_pack_cost_price
# Subtract total cost from revenue to isolate product-level profit for pen packs.
pen_pack_profit = pen_pack_revenue - pen_pack_total_cost
# Divide profit by revenue to express profitability as a percentage of sales.
pen_pack_profit_margin = (pen_pack_profit / pen_pack_revenue) * 100
# Subtract cost price from selling price to know the margin per individual pen pack.
pen_pack_profit_per_unit = pen_pack_selling_price - pen_pack_cost_price

# --- School bag ---
# Multiply units by selling price to calculate total sales income from school bags.
school_bag_revenue = school_bag_units * school_bag_selling_price
# Multiply units by cost price to calculate total direct inventory cost for school bags.
school_bag_total_cost = school_bag_units * school_bag_cost_price
# Subtract total cost from revenue to find total product profit from school bags.
school_bag_profit = school_bag_revenue - school_bag_total_cost
# Divide profit by revenue so margins can be compared fairly across products with different prices.
school_bag_profit_margin = (school_bag_profit / school_bag_revenue) * 100
# Subtract cost price from selling price to see the profit earned per school bag sold.
school_bag_profit_per_unit = school_bag_selling_price - school_bag_cost_price
```

#### Step 4: calculate totals


```python
# Sum all three product revenues to get the shop's total top-line sales.
total_revenue = notebook_revenue + pen_pack_revenue + school_bag_revenue
# Sum all three product costs to get the total Cost of Goods Sold (COGS).
total_cost = notebook_total_cost + pen_pack_total_cost + school_bag_total_cost

# Subtract total product cost from total revenue to find profit before fixed shop overhead.
gross_profit = total_revenue - total_cost
# Subtract fixed monthly expenses after gross profit because rent/bills must be paid regardless of product sales.
net_profit = gross_profit - monthly_expenses

# Divide gross profit by total_revenue (the denominator for all sales margins) to see efficiency before overhead.
gross_margin = (gross_profit / total_revenue) * 100
# Divide net profit by total_revenue to see how much of every rupee sold remains as true bottom-line profit.
net_margin = (net_profit / total_revenue) * 100

# ------------------------------------------
# Quick verification prints
# ------------------------------------------
print("--- Notebook ---")
print("Revenue:", notebook_revenue)
print("Total Cost:", notebook_total_cost)
print("Profit:", notebook_profit)
print("Profit Margin %:", notebook_profit_margin)
print("Profit Per Unit:", notebook_profit_per_unit)

print("\n--- Pen Pack ---")
print("Revenue:", pen_pack_revenue)
print("Total Cost:", pen_pack_total_cost)
print("Profit:", pen_pack_profit)
print("Profit Margin %:", pen_pack_profit_margin)
print("Profit Per Unit:", pen_pack_profit_per_unit)

print("\n--- School Bag ---")
print("Revenue:", school_bag_revenue)
print("Total Cost:", school_bag_total_cost)
print("Profit:", school_bag_profit)
print("Profit Margin %:", school_bag_profit_margin)
print("Profit Per Unit:", school_bag_profit_per_unit)

print("\n--- Overall Totals ---")
print("Total Revenue:", total_revenue)
print("Total Cost:", total_cost)
print("Gross Profit:", gross_profit)
print("Net Profit:", net_profit)
print("Gross Margin %:", gross_margin)
print("Net Margin %:", net_margin)
```

#### Step 5: comparisons

```python
# Business question: Was net profit at or above the target?
met_profit_target = net_profit >= monthly_profit_target

# Business question: Was the school bag's profit higher than the notebook's?
bag_beat_notebook = school_bag_profit > notebook_profit

# Business question: Were both of these true: the gross margin above 25% and the net profit at or above the target?
strong_performance = (gross_margin > 25) and (met_profit_target)

# Print results with labels
print("Net profit at or above target:", met_profit_target)
print("School bag profit higher than notebook profit:", bag_beat_notebook)
print("Gross margin above 25% AND net profit at/above target:", strong_performance)
```


#### Step 6: new product input

```python
# ------------------------------------------
# 1. Ask the shop owner for product details
# ------------------------------------------
# Keep product name as text (str) because a name is not used in math calculations.
new_product_name = input("Enter new product name: ")

# Collect numeric values as text first using clear prompts with units/currency.
new_product_units_text = input("Enter units sold: ")
new_product_selling_price_text = input("Selling price (₹): ")
new_product_cost_price_text = input("Cost price (₹): ")


# ------------------------------------------
# 2. Convert inputs to the right data types
# ------------------------------------------
# Convert units to int because physical items are sold in whole numbers (same rule as Part A).
new_product_units = int(new_product_units_text)

# Convert prices to float to support paise/decimals safely (same rule as Part A).
new_product_selling_price = float(new_product_selling_price_text)
new_product_cost_price = float(new_product_cost_price_text)


# ------------------------------------------
# 3. Calculate revenue, profit, and margin %
# ------------------------------------------
# Multiply units by selling price to find total sales earned for the new product.
new_product_revenue = new_product_units * new_product_selling_price

# Calculate total cost first, then subtract from revenue to find product profit.
new_product_total_cost = new_product_units * new_product_cost_price
new_product_profit = new_product_revenue - new_product_total_cost

# Divide profit by revenue and multiply by 100 to find the profit margin percentage.
new_product_margin = (new_product_profit / new_product_revenue) * 100


# ------------------------------------------
# 4. Print the three results with clear labels
# ------------------------------------------
print("\n--- New Product Summary:", new_product_name, "---")
print(new_product_name, "Revenue: ₹", new_product_revenue)
print(new_product_name, "Profit: ₹", new_product_profit)
print(new_product_name, "Profit Margin %:", new_product_margin)
```


#### Step 7: print report

```python
# ==========================================
# PART E: QUICKMART MONTHLY REPORT
# ==========================================

report_title = "======== QUICKMART MONTHLY REPORT ========"

# Title and matching divider lines
print("\n" + "=" * 42)
print(report_title)
print("=" * 42)

# --- Products Section ---
print("\n--- Products ---")

print("Notebook")
print(f"  Revenue: ₹{notebook_revenue:.2f} | Cost: ₹{notebook_total_cost:.2f}")
print(f"  Profit : ₹{notebook_profit:.2f} ({notebook_profit_margin:.2f}%, ₹{notebook_profit_per_unit:.2f}/unit)\n")

print("Pen Pack")
print(f"  Revenue: ₹{pen_pack_revenue:.2f} | Cost: ₹{pen_pack_total_cost:.2f}")
print(f"  Profit : ₹{pen_pack_profit:.2f} ({pen_pack_profit_margin:.2f}%, ₹{pen_pack_profit_per_unit:.2f}/unit)\n")

print("School Bag")
print(f"  Revenue: ₹{school_bag_revenue:.2f} | Cost: ₹{school_bag_total_cost:.2f}")
print(f"  Profit : ₹{school_bag_profit:.2f} ({school_bag_profit_margin:.2f}%, ₹{school_bag_profit_per_unit:.2f}/unit)")
print("-" * 42)

# --- Overall Section ---
print("\n--- Overall ---")
print(f"Total Revenue    : ₹{total_revenue:.2f}")
print(f"Total Cost       : ₹{total_cost:.2f}")
print(f"Gross Profit     : ₹{gross_profit:.2f} ({gross_margin:.2f}%)")
print(f"Monthly Expenses : ₹{monthly_expenses:.2f}")
print(f"Net Profit       : ₹{net_profit:.2f} ({net_margin:.2f}%)")
print("-" * 42)

# --- Targets Section ---
# Dynamically embed monthly_profit_target using an f-string (with comma formatting :,.0f)
print("\n--- Targets ---")
print(f"Net profit >= ₹{monthly_profit_target:,.0f} target     :", met_profit_target)
print("School bag beat notebook profit :", bag_beat_notebook)
print("Gross margin > 25% & target met :", strong_performance)
print("-" * 42)

# --- New Product Section ---
# Using end=" " and sep="" so two items share one clean line without extra spaces
print("\n--- New product ---")
print(f"Product : {new_product_name}", end=" ")
print(f"({new_product_units} units sold)")
print("Revenue : ₹", f"{new_product_revenue:.2f}", sep="")
print("Profit  : ₹", f"{new_product_profit:.2f}", sep="")
print(f"Margin  : {new_product_margin:.2f}%")
print("=" * 42)
```
