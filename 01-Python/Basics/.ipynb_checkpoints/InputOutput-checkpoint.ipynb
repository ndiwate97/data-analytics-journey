### 4. Ask for the number of units sold and the price per unit. Calculate the revenue and print it as Revenue: <value>.

```python
# Ask the user for the total units sold and convert the input to an integer
units_sold = int(input("Enter the number of units sold: "))

# Ask the user for the price of one unit and convert the input to a decimal number
price_per_unit = float(input("Enter the price per unit: "))

# Multiply units sold by price per unit to calculate the total revenue
revenue = units_sold * price_per_unit

# Display the final calculated revenue with the required label
print("Revenue:", revenue)
```

### 7. Ask for a product’s cost price and selling price. Print the profit and the profit margin % (profit ÷ selling price × 100), with a clear label for each.

```python
# Ask for the product's cost price and convert it to a decimal number
cost_price = float(input("Enter the cost price: "))

# Ask for the product's selling price and convert it to a decimal number
selling_price = float(input("Enter the selling price: "))

# Calculate the profit by subtracting cost price from selling price
profit = selling_price - cost_price

# Calculate the profit margin percentage based on the selling price
profit_margin = (profit / selling_price) * 100

# Print the profit and profit margin with clear labels
print(f"Profit: {profit:.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

### 8. Ask for minutes (a whole number). Print it as X hours Y minutes, using // and %.

```python
# Ask the user for the total number of minutes as a whole number
total_minutes = int(input("Enter minutes: "))

# Use floor division (//) to get the number of full hours
hours = total_minutes // 60

# Use the modulo operator (%) to get the remaining leftover minutes
minutes = total_minutes % 60

# Print the result in the format "X hours Y minutes"
print(f"{hours} hours {minutes} minutes")
```

### 9. A shopkeeper enters one sale at the till. Ask the user for:

Customer name,
Items bought (whole number),
Price per item (decimal),
Discount % (a number),

Then:

Convert each input to the right type.

Calculate the total before discount, the discount amount, and the final amount.

Print a short receipt using f-strings, with the sep or end you learned, so that it looks neat.

Add a comment on each calculation line.

```python
# Collect inputs and convert each to the proper data type
customer_name = input("Customer name: ")
items_bought = int(input("Items bought: "))
price_per_item = float(input("Price per item: "))
discount_percent = float(input("Discount %: "))

# Multiply items bought by price per item to get the subtotal
total_before_discount = items_bought * price_per_item

# Multiply the subtotal by the discount percentage divided by 100
discount_amount = total_before_discount * (discount_percent / 100)

# Subtract the discount amount from the subtotal to get the final bill
final_amount = total_before_discount - discount_amount

# Print a formatted receipt using f-strings, sep, and end
print("\n--- SALES RECEIPT ---")
print(f"Customer: {customer_name}", f"Items: {items_bought}", sep=" | ")
print(f"Subtotal: ₹{total_before_discount:.2f}", end="  |  ")
print(f"Discount ({discount_percent}%): -₹{discount_amount:.2f}")
print("-" * 21)
print(f"Final Amount: ₹{final_amount:.2f}")
```
