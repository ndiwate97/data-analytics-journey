### 5. Predict, then explain which part Python does first:

```python
print(True or False and False)      # Output: True ('and' has higher precedence than 'or', so it evaluates as True or (False and False) -> True or False)
print((True or False) and False)    # Output: False (Parentheses force 'or' to evaluate first, turning it into True and False)
```

### 6. price = "500" and budget = 500. What does price == budget give? Explain why, then write one line that compares them correctly. (This uses Day 3 ideas.)

```python
price = "500"                 # Stores "500" as a string (text) data type
budget = 500                  # Stores 500 as an integer (whole number)
print(price == budget)        # Output: False (Python does not treat a string and an integer as equal)
#fix
print(int(price) == budget)   # Output: True (int() converts the string "500" into the integer 500 before comparing)
```

### 8. Write a single expression for: “an order qualifies for free shipping if the amount is at least 999, or the customer is a member and the amount is at least 499”. Test it with three different sets of values and predict each result first.

```python
# Test Case 1: High amount, non-member
amount = 1000
is_member = False
print(amount >= 999 or (is_member and amount >= 499))  # Output: True

# Test Case 2: Medium amount, member
amount = 550
is_member = True
print(amount >= 999 or (is_member and amount >= 499))  # Output: True

# Test Case 3: Medium amount, non-member
amount = 600
is_member = False
print(amount >= 999 or (is_member and amount >= 499))  # Output: False
```

### 9. A store checks whether an order gets a bonus. Use these values (the first two arrive as text):

amount_text = "1800"
items_text = "6"
region = "West"
is_member = True

Using conversion, comparison and logical operators:

Convert the text values to the right types.
Check each condition on its own and print each answer with a label: amount above 1500, items at least 5, region is "West".
Print whether the order gets a bonus if all three are true.
Print whether it gets a discount if it is not a member or the amount is above 1500.

Add a comment on each line.

```python
amount_text = "1800"                                      # Stores the order amount as a string
items_text = "6"                                          # Stores the item count as a string
region = "West"                                           # Stores the customer's region as a string
is_member = True                                          # Stores the membership status as a boolean

amount = int(amount_text)                                 # Converts amount_text from a string to an integer (1800)
items = int(items_text)                                   # Converts items_text from a string to an integer (6)

above_1500 = amount > 1500                                # Checks if amount is strictly greater than 1500 (True)
at_least_5 = items >= 5                                   # Checks if items is greater than or equal to 5 (True)
is_west = region == "West"                                # Checks if region matches "West" exactly (True)

print("Amount above 1500:", above_1500)                   # Prints the label and result for the amount check
print("Items at least 5:", at_least_5)                    # Prints the label and result for the items check
print("Region is West:", is_west)                         # Prints the label and result for the region check

gets_bonus = above_1500 and at_least_5 and is_west        # Evaluates to True only if all three conditions are True
print("Gets bonus:", gets_bonus)                          # Prints whether the order qualifies for a bonus

gets_discount = (not is_member) or above_1500             # Evaluates to True if not a member OR amount is above 1500
print("Gets discount:", gets_discount)  
```
