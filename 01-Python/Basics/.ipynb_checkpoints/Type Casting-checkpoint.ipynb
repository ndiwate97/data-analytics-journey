### 9. A store’s system exports one order as text:

customer = "Rahul"
items_text = "3"
price_text = "799.50"
discount_text = "10" (a 10% discount)

Using conversion:

Convert the numbers to the correct types.
Calculate the total before discount, the discount amount, and the final amount.
Print the final amount as: Final amount: <value>

Add comments explaining each conversion.

```python
customer = "Rahul" 
items_text = "3" 
price_text = "799.50" 
discount_text = "10" # 10% discount

# Convert items_text to int because the number of items must be a whole number
items = int(items_text)

# Convert price_text to float because currency includes decimal values
price = float(price_text)

# Convert discount_text to float to allow precise percentage calculations
discount_pct = float(discount_text)


# Calculate the total cost before any discount is applied
total_before_discount = items * price

# Calculate the actual currency amount to subtract based on the percentage
discount_amount = total_before_discount * (discount_pct / 100)

# Subtract the discount amount from the initial total to get the final price
final_amount = total_before_discount - discount_amount

# Print the final formatted result
print(f"Final amount: {final_amount}")
```
