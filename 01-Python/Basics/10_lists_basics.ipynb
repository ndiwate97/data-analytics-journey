### 4. Calculate and print, with labels: total, average (2 decimals), highest and lowest sales.

```python
daily_sales = [1200, 1500, 1000, 1800, 2200]

total = sum(daily_sales)                      # Adds all numbers in the list together
average = total / len(daily_sales)            # Divides the total by the number of days (5)
highest = max(daily_sales)                    # Finds the largest number in the list
lowest = min(daily_sales)                     # Finds the smallest number in the list

print(f"Total sales: {total}")                # Prints the total sum with a descriptive label
print(f"Average sales: {average:.2f}")        # Prints the average formatted to 2 decimal places
print(f"Highest sales: {highest}")            # Prints the maximum value with a label
print(f"Lowest sales: {lowest}")              # Prints the minimum value with a label
```

### 7. Using slicing and sum(), compare the total of the first 3 days with the total of the last 2 days, and print True if the first 3 days earned more.

```python
daily_sales = [1200, 1500, 1000, 1800, 2200]  # Uses the updated daily sales list

first_3_total = sum(daily_sales[:3])          # Sums the first 3 days: 1200 + 1500 + 1000 = 3700
last_2_total = sum(daily_sales[-2:])          # Sums the last 2 days: 1800 + 2200 = 4000

print(first_3_total > last_2_total)           # Compares 3700 > 4000 and prints False
```

### 8. You have names = ["Notebook", "Pen", "Bag"] and prices = [45, 20, 650]. Print one line per product, like Notebook costs ₹45, using indexes (loops come later), and then print the total of all prices.

```python
names = ["Notebook", "Pen", "Bag"]            # Creates the list of product names
prices = [45, 20, 650]                        # Creates the list of matching product prices

print(f"{names[0]} costs ₹{prices[0]}")       # Pairs index 0 from both lists: Notebook costs ₹45
print(f"{names[1]} costs ₹{prices[1]}")       # Pairs index 1 from both lists: Pen costs ₹20
print(f"{names[2]} costs ₹{prices[2]}")       # Pairs index 2 from both lists: Bag costs ₹650

print(f"Total: ₹{sum(prices)}")               # Sums all values in the prices list (715)
```

### 9. A shop’s sales for one week (Mon to Sun):

##### week_sales = [2200, 1800, 2500, 2100, 3000, 4200, 3900]

##### Calculate the weekly total and the daily average.
##### Find the best and worst day’s sales.
##### Use slicing to get the weekday sales (Mon to Fri) and the weekend sales (Sat and Sun), and calculate the average of each.
##### Print whether the weekend average is more than 1.5 times the weekday average.
##### Print a neat report with f-strings, thousands commas and 2 decimals.

##### Add a comment on each line explaining why.

```python
week_sales = [2200, 1800, 2500, 2100, 3000, 4200, 3900]  # Store daily sales from Monday (index 0) to Sunday (index 6)

weekly_total = sum(week_sales)                           # Add all 7 days to get total weekly revenue
daily_avg = weekly_total / len(week_sales)               # Divide total revenue by 7 days to find the overall daily average
best_day = max(week_sales)                               # Find the highest single-day sales figure
worst_day = min(week_sales)                              # Find the lowest single-day sales figure

weekday_sales = week_sales[:5]                           # Slice the first 5 items (indices 0 to 4) for Mon to Fri
weekend_sales = week_sales[-2:]                          # Slice the last 2 items (indices 5 and 6) for Sat and Sun

weekday_avg = sum(weekday_sales) / len(weekday_sales)    # Calculate the average sales across the 5 weekdays
weekend_avg = sum(weekend_sales) / len(weekend_sales)    # Calculate the average sales across the 2 weekend days

is_weekend_strong = weekend_avg > (1.5 * weekday_avg)    # Check if weekend average exceeds 1.5 times the weekday average

print("----- Weekly Sales Report -----")                     # Print a clear header to organize the report
print(f"Weekly Total:      ₹{weekly_total:,.2f}")        # Format total with commas and 2 decimals for currency readability
print(f"Daily Average:     ₹{daily_avg:,.2f}")           # Format overall daily average consistently as currency
print(f"Best Day Sales:    ₹{best_day:,.2f}")            # Display peak daily revenue with thousands separator
print(f"Worst Day Sales:   ₹{worst_day:,.2f}")           # Display lowest daily revenue with thousands separator
print(f"Weekday Average:   ₹{weekday_avg:,.2f}")         # Show Mon–Fri average for comparison against the weekend
print(f"Weekend Average:   ₹{weekend_avg:,.2f}")         # Show Sat–Sun average for comparison against weekdays
print(f"Weekend > 1.5x Weekday: {is_weekend_strong}")    # Output True or False for the weekend performance check
```
