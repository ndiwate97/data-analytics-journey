### 8. A delivery takes 245 minutes. Write code that prints it as: 4 hours 5 minutes. Don’t hard-code the 4 and the 5.

```python
# --- Problem 8: Time Conversion ---
total_minutes = 245

# Calculate hours using floor division and remaining minutes using modulo
hours = total_minutes // 60
minutes = total_minutes % 60

# Print the formatted time duration
print(f"{hours} hours {minutes} minutes")
```

### 9. A store’s export gives these values as text:

units_text = "200"
price = 400
cost_per_unit = 300

Using conversion and operators:

Convert units_text to the right type.
Calculate the revenue, total cost and profit.
Calculate the profit margin as a percentage (profit ÷ revenue × 100) and the profit per unit.
Print each result with a clear label.

Add a comment on each calculation line.

```python
# --- Store Export Analysis ---
units_text = "200"
price = 400
cost_per_unit = 300

# Convert text-based units into an integer for numerical operations
units = int(units_text)

# Calculate total revenue by multiplying units sold by the selling price
revenue = units * price

# Calculate total cost by multiplying units by the cost per unit
total_cost = units * cost_per_unit

# Calculate total profit by subtracting total cost from total revenue
profit = revenue - total_cost

# Calculate profit margin percentage (profit divided by revenue multiplied by 100)
profit_margin_pct = (profit / revenue) * 100

# Calculate the profit made on each individual unit
profit_per_unit = price - cost_per_unit

# Print the financial metrics with clear, readable labels
print(f"Revenue: ₹{revenue}")
print(f"Total Cost: ₹{total_cost}")
print(f"Total Profit: ₹{profit}")
print(f"Profit Margin: {profit_margin_pct}%")
print(f"Profit per Unit: ₹{profit_per_unit}")
```
