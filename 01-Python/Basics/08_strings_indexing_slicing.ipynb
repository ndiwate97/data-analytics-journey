### 5. Predict each output:


```python
s = "ANALYTICS"
print(s[:4])      #--------->First four characters. here the start index begins at index 0 (A) and stops right before index 4 (Y), returning ANAL
print(s[4:])      #--------->Starting at index 4 (Y) and omitting the stop index grabs everything till the end of the string, returning YTICS.
print(s[::-1])    #--------->Reversed string. Using a negative step of -1 walks backward through the string from end to start, returning SCITYLANA.
print(s[::2])     #--------->Every second character. Using a step of 2 starts at index 0 (A) and skips every other character, picking indices 0, 2, 4, 6, 8 to return AAYIS.
```

### 9. An online store stores each order ID as text: order_id = "WEST-2026-00458".

Extract the region, the year and the serial number.

Convert the year and serial to integers, and calculate the next order number.

Print: Order 458 from WEST placed in 2026 (using your variables, not typed numbers).

Print how many characters the ID has.

In a comment, explain why it makes sense to store order_id as text, not a number. (Think back to Day 2, Q8.)

Add a comment on each line explaining why.

```python
# Store the order ID as a string so letters, hyphens, and leading zeros stay intact
order_id = "WEST-2026-00458"

# Slice from index 0 up to 4 to grab the region code ("WEST")
region = order_id[0:4]

# Slice from index 5 up to 9 to extract the year text ("2026")
year_text = order_id[5:9]

# Slice from index 10 to the end to extract the serial text ("00458")
serial_text = order_id[10:]

# Convert year_text to an integer so Python treats it as a real number
year = int(year_text)

# Convert serial_text to an integer (which drops leading zeros to give 458) so we can do math
serial = int(serial_text)

# Add 1 to the serial integer to calculate the next sequential order number (459)
next_order = serial + 1
print("Next order number:", next_order)

# Use an f-string to insert our extracted variables directly into the required sentence
print(f"Order {serial} from {region} placed in {year}")

# Use len() to count and print the total number of characters (including hyphens) in order_id
print(len(order_id))

# Why store order_id as text instead of a number?
# 1. It contains letters ("WEST") and hyphens ("-"), which cannot exist in an integer or float.
# 2. Even for pure digits, numbers drop leading zeros (turning "00458" into 458), and IDs are labels rather than quantities we do math on.
```
