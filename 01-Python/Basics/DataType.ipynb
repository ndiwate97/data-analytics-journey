### 9. An online store has this order:

Customer: Rahul

Items ordered: 3

Price per item: ₹799.50

Used a discount coupon: Yes

Using variables with the correct data types:

Create a variable for each detail.
Print each value along with its type.
Add a comment above each variable explaining why you chose that type.

```python
# Since name is a text hence string 
customer = "Rahul" 
# item order will be a whole no
items_ordered = 3 
# price can be in decimal no, hence float
price = 799.50
# Used a discount coupon will have 2 value either yes or no, hence bool
is_discount_coupon = True


print("Customer:",customer, "type =",type(customer))
print("Items ordered:",items_ordered, "type =",type(items_ordered))
print("Price per item:",price, "type =",type(price))
print("Used a discount coupon:",is_discount_coupon, "type =",type(is_discount_coupon))
```
