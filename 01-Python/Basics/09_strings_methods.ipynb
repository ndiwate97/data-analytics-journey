### 6. Two region values arrive as "West" and " WEST ". Write code that proves they’re the same region, and print True. Then explain why a plain == gives False.

### 7. price_text = " ₹1,299.50 ". Convert it to a float, add 10%, and print the result with 2 decimals and thousands commas. (Day 3, Day 4 and the formatting table.)

```python
price_text = " ₹1,299.50 "

# Clean spaces, currency symbol, and comma, then convert to float
price = float(price_text.strip().replace("₹", "").replace(",", ""))

# Add 10%
total = price * 1.10

# Format with thousands commas and 2 decimal places
print(f"{total:,.2f}")
```

### 9. Predict each output, then run to check:

```python
x = 1234567.891
print(f"{x:.2f}")    # Output: 1234567.89 (rounds to 2 decimal places, no commas)
print(f"{x:,}")      # Output: 1,234,567.891 (adds thousands commas, keeps all decimals)
print(f"{x:,.2f}")   # Output: 1,234,567.89 (adds thousands commas AND rounds to 2 decimals)
```

### 10. Write the same sentence three ways (f-string, .format(), %): Riya spent 1200 rupees

```python
name = "Riya"
amount = 1200

# 1. f-string (modern and most readable)
print(f"{name} spent {amount} rupees")

# 2. .format() method
print("{} spent {} rupees".format(name, amount))

# 3. % formatting (legacy style)
print("%s spent %d rupees" % (name, amount))
```

### 11. A store exports one messy order:
##### name = " rIYA sHAH "
##### region = " west "
##### amount_text = " ₹12,500 "
##### order_id = "west-2026-00458"
##### Clean the name (proper case), the region (upper case), and the order ID (upper case).
##### Convert amount_text to a number.
##### Check that the cleaned order_id starts with the cleaned region, and print the result.
##### Calculate 18% GST and the final total.
##### Print a neat receipt using f-strings, with thousands commas and 2 decimals.


```python
# Store the raw, uncleaned export values from the store
name = " rIYA sHAH "
region = " west "
amount_text = " ₹12,500 "
order_id = "west-2026-00458"

# Remove outer spaces with .strip() and capitalize each word with .title()
clean_name = name.strip().title()

# Remove outer spaces and convert to uppercase so region codes stay uniform
clean_region = region.strip().upper()

# Remove any accidental spaces and convert to uppercase to standardize the ID
clean_order_id = order_id.strip().upper()

# Strip spaces, remove "₹" and ",", and convert to float so we can run tax math
amount = float(amount_text.strip().replace("₹", "").replace(",", ""))

# Verify that the order ID starts with the matching region code (returns True)
is_valid_region = clean_order_id.startswith(clean_region)

# Print the validation result to confirm the data is consistent
print("Region matches Order ID:", is_valid_region)

# Calculate 18% GST by multiplying the base amount by 0.18
gst = amount * 0.18

# Add the base amount and GST together to get the final payable total
total = amount + gst

# Print a divider line of 30 dashes to visually separate the receipt
print("-" * 30)

# Display the cleaned customer and order details using f-strings
print(f"Order ID : {clean_order_id}")
print(f"Customer : {clean_name}")
print(f"Region   : {clean_region}")

# Print another divider line before the financial breakdown
print("-" * 30)

# Format all currency figures with thousands commas (,) and 2 decimal places (.2f)
print(f"Subtotal : ₹{amount:,.2f}")
print(f"GST (18%): ₹{gst:,.2f}")
print(f"Total    : ₹{total:,.2f}")

# Print a closing divider line at the bottom of the receipt
print("-" * 30)
```
